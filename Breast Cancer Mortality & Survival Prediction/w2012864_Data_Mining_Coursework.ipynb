#Data Pre-processing

##One Hot Encode Nominal Variables and Label Encode Ordinal Variables

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, LabelEncoder
from sklearn.compose import ColumnTransformer
```

```python
nominal_cols = ['Sex', 'Estrogen_Status', 'Progesterone_Status']
ordinal_cols = ['T_Stage', 'N_Stage', '6th_Stage', 'Differentiated', 'A_Stage']
numerical_cols = ['Age', 'Grade', 'Tumor_Size', 'Regional_Node_Examined', 'Reginol_Node_Positive', 'Survival_Months']
```

```python
t_stage_order = ['T1', 'T2', 'T3', 'T4']
n_stage_order = ['N1', 'N2', 'N3']
sixth_stage_order = ['IIA', 'IIB', 'IIIA', 'IIIB', 'IIIC']
differentiated_order = ['Well differentiated', 'Moderately differentiated', 'Poorly differentiated','Undifferentiated'] # Example, confirm actual unique values and order
a_stage_order = ['Regional', 'Distant']
```

```python
all_ordinal_categories = [t_stage_order, n_stage_order, sixth_stage_order, differentiated_order, a_stage_order]
```

```python
feature_preprocessor = ColumnTransformer(
    transformers=[
        ('onehot', OneHotEncoder(drop='first', handle_unknown='ignore'), nominal_cols),
        ('ordinal', OrdinalEncoder(categories=all_ordinal_categories, handle_unknown='use_encoded_value', unknown_value=-1), ordinal_cols),
        ('numerical_passthrough', 'passthrough', numerical_cols)
    ],
    remainder='drop'
)
```

```python
tree = dt2_regressor.tree_
```

```python
node_indicator = dt2_regressor.decision_path(patient_processed_df)
```

```python
leaf_id = dt2_regressor.apply(patient_processed_df)
```

```python
feature_names_dt = X_regression_processed_df.columns.tolist()
```

```python
current_node_id = 0
path_rules = []

while True:
    if tree.children_left[current_node_id] == tree.children_right[current_node_id]:
        path_rules.append(f"Reached final prediction: {tree.value[current_node_id][0][0]:.2f} months.")
        break

    feature_index = tree.feature[current_node_id]
    threshold = tree.threshold[current_node_id]
    feature_name = feature_names_dt[feature_index]
    patient_value = patient_processed_df.iloc[0, feature_index]

    if patient_value <= threshold:
        path_rules.append(f"Test: '{feature_name}' (value: {patient_value:.2f}) <= {threshold:.2f} (Go Left)")
        current_node_id = tree.children_left[current_node_id]
    else:
        path_rules.append(f"Test: '{feature_name}' (value: {patient_value:.2f}) > {threshold:.2f} (Go Right)")
        current_node_id = tree.children_right[current_node_id]

    if current_node_id == leaf_id[0]:
        path_rules.append(f"Reached final prediction: {tree.value[current_node_id][0][0]:.2f} months.")
        break

for step in path_rules:
  print(step)


```
